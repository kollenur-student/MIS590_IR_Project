# 🌊 Surfability thresholds explained: what score do you get?

This notebook explains every row of `data/resources/surfability_thresholds.csv` and shows, **for each skill level (beginner, intermediate, advanced), which score a given condition produces**.

**How the numbers were made.** Every score below was calculated with the project's real scoring function (`factor_scores` in `03_surfability.ipynb`), using the current CSV values. Nothing was computed by hand. If you change the CSV, these tables are no longer exact.

**The test spot.** All tables use a simple imaginary spot:
- it faces east (90°);
- it accepts swell from 60°–160°;
- it prefers mid tide;
- its profile lists all three skill levels.

In each table only **one** condition changes; everything else is ideal.

> **How to read the scores:** every factor is a number from **0** (kills the session) to **1** (perfect). The condition-fit score multiplies all factors:
>
> $$CF = \text{size} \times \text{period} \times \text{direction} \times \text{wind} \times \text{tide}^{0.5} \times \text{level}$$
>
> The verdict follows from CF: **Poor** below 0.5, **Fair** from 0.5, **Good** from 0.75.

## 1️⃣ Wave size: `wave_min_m`, `wave_ideal_low_m`, `wave_ideal_high_m`, `wave_max_m`

### Thresholds

| Parameter | Beginner | Intermediate | Advanced | Meaning |
|---|---|---|---|---|
| `wave_min_m` | 0.3 m | 0.6 m | 1 m | Below this: too small → score **0** |
| `wave_ideal_low_m` | 0.3 m | 1 m | 1.5 m | Score reaches **1** |
| `wave_ideal_high_m` | 0.6 m | 1.5 m | 3 m | Score stays **1** up to here |
| `wave_max_m` | 1 m | 1.8 m | 4 m | Above this: **unsafe** → score 0 and a safety warning |

### Rule

The size score is a trapezoid:
- **0** at or below `wave_min_m`;
- rising in a straight line to **1** at `wave_ideal_low_m`;
- **1** up to `wave_ideal_high_m`;
- falling in a straight line to **0** at `wave_max_m`.

Example for intermediate, 0.8 m: (0.8 − 0.6) / (1.0 − 0.6) = **0.50**.

### What this means

- **Beginners** get full points from 0.3 to 0.6 m. At 0.8 m they get 0.5, and from 1.0 m the score is 0 (unsafe above 1.0 m). Their minimum and ideal-low are both 0.3 m, so the score jumps straight from 0 to 1 at 0.3 m.
- **Intermediates** score 0 below 0.6 m. They have full points from 1.0 to 1.5 m, and are unsafe above 1.8 m.
- **Advanced** surfers score 0 below 1.0 m, have full points from 1.5 to 3.0 m, and are unsafe above 4.0 m.

## 2️⃣ Period: `period_min_s`, `period_ideal_s`, `period_floor_fit`

### Thresholds

| Parameter | Beginner | Intermediate | Advanced | Meaning |
|---|---|---|---|---|
| `period_min_s` | 5 s | 6 s | 7 s | At or below this: short, choppy wind swell |
| `period_ideal_s` | 8 s | 9 s | 10 s | Score reaches **1** (organised swell) |
| `period_floor_fit` | 0.5 | 0.4 | 0.3 | Minimum score for very short periods |

### Rule

- At or below `period_min_s`, the score is `period_floor_fit`. It never drops to 0, because short-period waves are weak but still surfable.
- It rises in a straight line to **1** at `period_ideal_s`.
- It stays 1 above that.

Example for intermediate, 7 s: 0.4 + (7 − 6) / (9 − 6) × (1 − 0.4) = **0.60**.

### Score per period

| Period | Beginner | Intermediate | Advanced |
|---|---|---|---|
| 3 s | 0.50 | 0.40 | 0.30 |
| 4 s | 0.50 | 0.40 | 0.30 |
| 5 s | 0.50 | 0.40 | 0.30 |
| 6 s | 0.67 | 0.40 | 0.30 |
| 7 s | 0.83 | 0.60 | 0.30 |
| 8 s | **1.00** | 0.80 | 0.53 |
| 9 s | **1.00** | **1.00** | 0.77 |
| 10 s | **1.00** | **1.00** | **1.00** |
| 12 s | **1.00** | **1.00** | **1.00** |
| 15 s | **1.00** | **1.00** | **1.00** |

### What this means

Periods of 4–6 s are common in Taiwan. At 5 s every level gets only its floor value: **0.5 / 0.4 / 0.3**. That is why explanations often say *"Limited by short period"*. Beginners are least affected, because they learn in the white water.

## 3️⃣ Swell direction: `swell_dir_min_deg`, `swell_dir_max_deg` (spot profile) and `swell_window_margin_deg`

### Thresholds

| Parameter | Value (all levels) | Meaning |
|---|---|---|
| swell window | from the spot profile (test spot: 60°–160°) | Directions the swell can come from to reach this beach |
| `swell_window_margin_deg` | 30° | Outside the window, the score falls from 1 to 0 over this many degrees |

### Rule

- Inside the window, the score is **1**.
- Outside, the score is 1 − (degrees outside ÷ 30).
- From 30° outside, the score is **0**.

The rule is the same for all levels.

### Score per swell direction (test spot window 60°–160°)

| Swell direction | Degrees outside the window | Score (all levels) |
|---|---|---|
| 90° | 0° | **1.00** |
| 160° | 0° | **1.00** |
| 165° | 5° | 0.83 |
| 170° | 10° | 0.67 |
| 175° | 15° | 0.50 |
| 180° | 20° | 0.33 |
| 185° | 25° | 0.17 |
| 190° | 30° | **0.00** |
| 200° | 40° | **0.00** |
| 220° | 60° | **0.00** |

### What this means

Swell from slightly outside the window still reaches the beach, but weaker, because waves bend around headlands. Swell more than 30° outside the window is treated as **not reaching the beach at all**: the score is 0, so the whole condition fit is 0.

## 4️⃣ Wind: `wind_glassy_max_kmh`, `wind_offshore_max_kmh`, `wind_offshore_limit_kmh`, `wind_onshore_max_kmh`

### Thresholds

| Parameter | Beginner | Intermediate | Advanced | Meaning |
|---|---|---|---|---|
| `wind_glassy_max_kmh` | 5.5 | 5.5 | 5.5 | Up to this speed: light/glassy → **1** from any direction |
| `wind_offshore_max_kmh` | 18.5 | 28 | 28 | Offshore wind up to this speed is ideal → **1** |
| `wind_offshore_limit_kmh` | 37 | 46 | 46 | Offshore wind this strong → **0** (too hard to paddle into waves) |
| `wind_onshore_max_kmh` | 28 | 28 | 28 | Straight onshore wind this strong → **0** |

All speeds are in km/h; 1 knot = 1.852 km/h.

### Rule

The direction of the wind relative to the beach decides which rule applies. The test beach faces east, so offshore wind comes **from the west** (270°).

- **Offshore** (from the land, within ±60° of 270°): score 1 up to `wind_offshore_max_kmh`, then falling to 0 at `wind_offshore_limit_kmh`.
- **Onshore and cross-shore:** the score falls from 1 (at the glassy limit) towards 0 (at `wind_onshore_max_kmh`). The penalty is scaled by how onshore the wind is:
  - straight onshore gets the full penalty;
  - exactly cross-shore (90° to the beach) gets one third of it.

## 5️⃣ Tide: `tide_mismatch_fit` and `weight_tide`

### Thresholds

| Parameter | Value (all levels) | Meaning |
|---|---|---|
| `tide_mismatch_fit` | 0.6 | Tide score when the tide stage is not one the spot prefers |
| `weight_tide` | 0.5 | Exponent: the tide counts only half as much as the other factors |

### Rule

- The tide stage (low / mid / high) is calculated from the forecast sea level.
- If it matches the spot's `ideal_tide`, or the spot accepts `any` tide, the score is **1**. Otherwise it is **0.6**.
- Because of the weight 0.5, the value enters the product as its square root: $0.6^{0.5} \approx 0.77$.

### Score

| Tide stage vs. spot preference | Tide factor | Effect in the product (factor^0.5) |
|---|---|---|
| matches (spot likes *mid*, tide is mid) | 1.00 | 1.00 |
| does not match (spot likes *mid*, tide is low) | 0.60 | 0.77 |
| spot accepts *any* tide | 1.00 | 1.00 |

### What this means

A wrong tide never kills a session. In the end it costs about **23 %** of the condition fit, for every skill level.

## 6️⃣ Spot level: `spot_skill_mismatch_fit`

### Threshold

| Parameter | Beginner | Intermediate | Advanced | Meaning |
|---|---|---|---|---|
| `spot_skill_mismatch_fit` | **0** | 0.4 | 0.7 | Score when the spot profile does **not** list your level |

### Rule

- If the spot's `skill_levels` in `spots.csv` include your level, the score is **1**.
- If not, the score is the value above.

### Score for every combination of user level and spot profile

| Spot profile lists … | Beginner user | Intermediate user | Advanced user |
|---|---|---|---|
| beginner | **1.00** | 0.40 | 0.70 |
| beginner, intermediate | **1.00** | **1.00** | 0.70 |
| beginner, intermediate, advanced | **1.00** | **1.00** | **1.00** |
| intermediate, advanced | **0.00** 🚫 | **1.00** | **1.00** |
| advanced | **0.00** 🚫 | 0.40 | **1.00** |

🚫 = score 0. The spot is **never recommended** to this user: it drops to the bottom of the ranking with a safety warning.

### What this means

- **Beginners** are protected: a spot that is not listed for beginners always scores 0 for them.
- **Intermediates** at a beginner-only or advanced-only spot keep 40 %.
- **Advanced** surfers at a spot below their level keep 70 %; they can surf there, it is just less interesting.

## 7️⃣ Weights: `weight_size`, `weight_period`, `weight_direction`, `weight_wind`, `weight_tide`

| Parameter | Value (all levels) | Effect |
|---|---|---|
| `weight_size` | 1 | Size factor counts fully |
| `weight_period` | 1 | Period factor counts fully |
| `weight_direction` | 1 | Direction factor counts fully |
| `weight_wind` | 1 | Wind factor counts fully |
| `weight_tide` | 0.5 | Tide factor counts half (square root) |

The weights are **exponents** in the product. A factor of 0.5 with weight 1 stays 0.5. With weight 0.5 it becomes $0.5^{0.5} \approx 0.71$, so it pulls the score down less. A weight of 2 would make a factor twice as decisive.

## 8️⃣ Verdicts and daylight: `surfable_threshold`, `good_threshold`, `first_light_hour`, `last_light_hour`

| Parameter | Value (all levels) | Meaning |
|---|---|---|
| `surfable_threshold` | 0.5 | CF from this value = **Fair** (counts as surfable) |
| `good_threshold` | 0.75 | CF from this value = **Good** |
| `first_light_hour` | 06:00 | First hour considered |
| `last_light_hour` | 18:00 | Hours from this one on are not considered |

| Condition fit (CF) | Verdict |
|---|---|
| below 0.5 | **Poor** |
| 0.5 to below 0.75 | **Fair** (surfable) |
| 0.75 and above | **Good** |

Only daylight hours between 06:00 and 18:00 are scored. The best of those hours in your time window counts.

### Key takeaways

- **One weak factor dominates.** Because the factors are multiplied, a single 0 (too small, too big, swell from the wrong side, strong onshore wind) makes the whole CF 0.
- **The same waves mean different things per level.** Clean 1.2 m groundswell is **Good** for intermediates, **Poor** for beginners (too big, unsafe) and **Poor** for advanced surfers (too small).
- **Short periods hold Taiwan's scores down.** At 4–6 s every level keeps at most 0.5 of the period factor, so a "Good" verdict needs organised swell (8–10 s and more).
